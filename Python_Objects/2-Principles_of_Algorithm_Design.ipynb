Algorithm design is the systematic process of creating efficient, scalable, and provably correct step-by-step procedures to solve computational problems. Designing robust algorithms involves selecting the right paradigm, analyzing resource bounds (time and space), and handling structural trade-offs.

---

# **1. Core Principles & Design Paradigms**

```
                      ┌─────────────────────────┐
                      │    Algorithm Design     │
                      └────────────┬────────────┘
                                   │
     ┌─────────────────┬───────────┴───────────┬─────────────────┐
     ▼                 ▼                       ▼                 ▼
┌─────────┐   ┌─────────────────┐    ┌──────────────────┐   ┌─────────┐
│ Greedy  │   │ Divide & Conquer│    │ Dynamic Prog.    │   │Backtrack│
└─────────┘   └─────────────────┘    └──────────────────┘   └─────────┘
```

### **1. Divide and Conquer**

- Breaks a problem into independent subproblems, recursively solves them, and combines their solutions.
    * **Mechanism:** $T(n) = aT(n/b) + f(n)$ (analyzed using the **Master Theorem**).
    * **Key Condition:** Subproblems must be disjoint and non-overlapping.
    * **Classic Examples:** Merge Sort, Quick Sort, Binary Search, Strassen's Matrix Multiplication.

### **2. Dynamic Programming (DP)**

Solves complex problems by breaking them down into **overlapping subproblems** and storing intermediate results (memoization or tabulation) to avoid redundant work.

* **Core Requirements:**
    - 1. **Optimal Substructure:** An optimal solution to the problem contains optimal solutions to its subproblems.
    - 2. **Overlapping Subproblems:** The recursive space contains repeated calls to the same subproblem.


* **Classic Examples:** 0/1 Knapsack, Longest Common Subsequence (LCS), Bellman-Ford Shortest Path, Floyd-Warshall.

### **3. Greedy Choice Paradigm**

Makes the locally optimal choice at each step with the hope of reaching a globally optimal solution.

* **Core Requirements:**
    - 1. **Greedy Choice Property:** A global optimum can be arrived at by making locally optimal choices without looking ahead or backtracking.
    - 2. **Optimal Substructure.**


* **Classic Examples:** Dijkstra’s Shortest Path, Huffman Coding, Kruskal’s & Prim’s Minimum Spanning Tree (MST), Fractional Knapsack.

### **4. Backtracking & Branch and Bound**

Systematically explores a state-space tree of candidate solutions, **pruning** paths that violate constraints early.

* **Backtracking (DFS-based):** Used for decision, enumeration, or search problems (e.g., $N$-Queens, Sudoku, Hamiltonian Path).
* **Branch and Bound (BFS/Priority-Queue-based):** Used for optimization problems with bounding functions (e.g., Travelling Salesperson Problem, Integer Linear Programming).

# **2. Fundamental Steps in Designing an Algorithm**

```
  1. Formalize Inputs/Outputs & Constraints
                    │
                    ▼
  2. Identify Structural Invariants & Paradigm
                    │
                    ▼
  3. Formulate Pseudocode & Step Sequence
                    │
                    ▼
  4. Prove Correctness (Induction / Loop Invariants)
                    │
                    ▼
  5. Analyze Asymptotic Complexity (Big-O / Big-Ω)
                    │
                    ▼
  6. Optimize & Code Implementation

```

1. **Problem Formulation:** Formally state the inputs, expected outputs, invariants, and edge cases (e.g., negative weights, duplicate keys).
2. **Invariant Definition:** Establish **loop invariants** or mathematical properties that must hold true before and after each iteration to guarantee correctness.
3. **Correctness Proof:** Prove that the algorithm terminates and returns the correct result using:
    * **Mathematical Induction:** Standard for recursive/Divide & Conquer strategies.
    * **Exchange Arguments:** Standard for proving Greedy algorithm optimality.
    * **Loop Invariants:** Standard for iterative loops.
4. **Complexity Analysis:** Evaluate worst-case, average-case, and best-case performance in terms of input size $n$.

    * **Space Complexity:** Auxiliary memory required beyond the input size.

- In many designs, you can sacrifice memory to reduce execution time (or vice versa):
    * **Precomputation & Lookup Tables:** E.g., Frequency arrays, Hash Maps, Prefix Sums ($O(1)$ query time after $O(n)$ space).
    * **Memoization Tables:** Trading $O(n)$ memory for an exponential reduction in recursive call trees ($O(2^n) \to O(n)$).

# **5. Modern Practical Considerations**

1. **Cache Locality:** Contiguous memory access (e.g., standard arrays) outperforms pointer-heavy data structures (e.g., linked lists or trees) due to CPU L1/L2 cache prefetching, even if theoretical $O(1)$ access is identical.
2. **Amortized Complexity:** Evaluating average time per operation over a sequence of operations (e.g., dynamic array resizing takes $O(1)$ amortized time).
3. **Randomization:** Utilizing random choices to avoid worst-case deterministic behavior (e.g., Randomized QuickSort avoiding $O(n^2)$, Treaps, Bloom Filters).

**Recursion** and **Backtracking** are fundamental algorithmic techniques used to solve complex problems by breaking them down into smaller subproblems or exploring potential decision spaces.

# **1. Recursion: The Core Foundation**

Recursion is a programming technique where a function **calls itself** to solve a smaller instance of the same problem.

- Every valid recursive function must have two components:
    - 1. **Base Case:** The condition under which the recursion stops. Without it, the function calls itself indefinitely, causing a **Stack Overflow**.
    - 2. **Recursive Step:** The logic that reduces the problem instance and calls the function with a smaller or simpler input.

```
                  ┌────────────────────────┐
                  │    factorial(3)        │  3 * factorial(2)
                  └───────────┬────────────┘
                              │
                  ┌───────────▼────────────┐
                  │    factorial(2)        │  2 * factorial(1)
                  └───────────┬────────────┘
                              │
                  ┌───────────▼────────────┐
                  │    factorial(1)        │  Base Case -> returns 1
                  └────────────────────────┘

```

### **Classic Example: Factorial in Python**

```python
def factorial(n: int) -> int:
    # 1. Base Case
    if n <= 1:
        return 1

    # 2. Recursive Step
    return n * factorial(n - 1)

```

---

# **2. Backtracking: Controlled Brute-Force with Pruning**

Backtracking is an algorithmic paradigm that builds candidate solutions incrementally and **abandons a candidate ("backtracks") as soon as it determines that the candidate cannot lead to a valid solution**.

It models the decision-making process as a **State-Space Tree** using **Depth-First Search (DFS)**.

```
                           Root (Empty State)
                         /        |        \
                   Choice A    Choice B   Choice C
                     /    \        |
                Choice A1 [X]     Valid   (Pruned/Backtracked [X])
```

### **The Universal Backtracking Blueprint**

- Most backtracking algorithms follow this exact structural pattern:

```python
def backtrack(state, choices, result):
    # 1. Base Case: Check if state is a valid complete solution
    if is_solution(state):
        result.append(state.copy())
        return

    for choice in choices:
        # 2. Pruning / Constraint Check
        if is_valid(choice, state):
            # 3. Make Choice
            state.append(choice)

            # 4. Recurse (Explore deeper state space)
            backtrack(state, choices, result)

            # 5. Backtrack (Undo Choice)
            state.pop()

```

---

# **3. Concrete Example: Generating Subsets (Power Set)**

- Given an array of unique integers `nums`, return all possible subsets.

```python
def subsets(nums: list[int]) -> list[list[int]]:
    result = []

    def backtrack(start_index: int, current_path: list[int]):
        # Add a copy of current subset at every node in the state-space tree
        result.append(current_path.copy())

        for i in range(start_index, len(nums)):
            # 1. Make choice
            current_path.append(nums[i])

            # 2. Explore
            backtrack(i + 1, current_path)

            # 3. Backtrack (Undo choice)
            current_path.pop()

    backtrack(0, [])
    return result


# Example Run:
print(subsets([1, 2]))
# Output: [[], [1], [1, 2], [2]]

```

# **4. Key Differences: Recursion vs. Backtracking**

| Feature | Recursion | Backtracking |
| --- | --- | --- |
| **Core Concept** | Solving a problem by executing subproblems of the same structure. | Systematically searching all solutions via DFS tree exploration. |
| **State Exploration** | Single or multiple linear paths (e.g., Divide & Conquer). | Explores multiple candidate branches and **prunes invalid ones**. |
| **Key Action** | Function calling itself. | **Make choice $\to$ Recurse $\to$ Undo choice**. |
| **Applications** | Merge Sort, Tree Traversals, Factorial, Fibonacci. | $N$-Queens, Sudoku Solver, Subset/Permutation Generation, Graph Coloration. |

# **5. Time & Space Complexity**

* **Time Complexity:** Usually exponential or factorial, as backtracking systematically explores state-space trees:
    * Subsets / Combination problems: **$\mathcal{O}(2^n)$**
    * Permutations / $N$-Queens problems: **$\mathcal{O}(n!)$**
* **Space Complexity:** Determined by the maximum depth of the recursive call stack: **$\mathcal{O}(d)$**, where $d$ is the maximum tree depth.

The **Karatsuba algorithm** is a fast multiplication algorithm discovered by Anatoly Karatsuba in 1960. It uses a **Divide and Conquer** strategy to multiply two $n$-digit numbers in $\mathcal{O}(n^{\log_2 3}) \approx \mathcal{O}(n^{1.585})$ time, significantly faster than the traditional grade-school multiplication algorithm which runs in $\mathcal{O}(n^2)$ time.

---

# **1. Mathematical Foundation**

- Standard multiplication splits two $n$-digit numbers $x$ and $y$ into higher and lower halves using a base $B$ (typically $10$ or $2$):
    - $$x = x_1 \cdot B^m + x_0$$
    - $$y = y_1 \cdot B^m + y_0$$

*(where $m = \lfloor n / 2 \rfloor$)*

- Multiplying $x$ and $y$ directly yields:

    - $$x \cdot y = (x_1 \cdot B^m + x_0)(y_1 \cdot B^m + y_0) = x_1 y_1 \cdot B^{2m} + (x_1 y_0 + x_0 y_1) \cdot B^m + x_0 y_0$$
    - Naively, computing $x \cdot y$ requires **4 multiplications**: $x_1 y_1$, $x_1 y_0$, $x_0 y_1$, and $x_0 y_0$. Recurrence: $T(n) = 4T(n/2) + \mathcal{O}(n) \implies \mathcal{O}(n^2)$.

### **The Karatsuba Trick**

- Karatsuba realized that the middle coefficient $(x_1 y_0 + x_0 y_1)$ can be computed using **only 1 additional multiplication** instead of 2 by leveraging algebraic identity:
    1. Compute $P_1 = x_1 \cdot y_1$
    2. Compute $P_2 = x_0 \cdot y_0$
    3. Compute $P_3 = (x_1 + x_0) \cdot (y_1 + y_0)$
    4. Express middle term: $(x_1 y_0 + x_0 y_1) = P_3 - P_1 - P_2$

- The full product becomes:
    - $$x \cdot y = P_1 \cdot B^{2m} + (P_3 - P_1 - P_2) \cdot B^m + P_2$$

- By reducing the recursive multiplications from 4 to 3, the recurrence relation becomes:
    - $$T(n) = 3T(n/2) + \mathcal{O}(n)$$

> By Case 1 of the **Master Theorem**, this yields a time complexity of **$\mathcal{O}(n^{\log_2 3}) \approx \mathcal{O}(n^{1.585})$**.

---

# **2. Step-by-Step Example**

Multiply $x = 1234$ and $y = 5678$ ($n = 4$, $B = 10$, $m = 2$).

1. **Split into halves:**
    * $x_1 = 12$, $x_0 = 34$
    * $y_1 = 56$, $y_0 = 78$


2. **Compute 3 multiplications ($P_1, P_2, P_3$):**
    * $P_1 = x_1 \cdot y_1 = 12 \cdot 56 = 672$
    * $P_2 = x_0 \cdot y_0 = 34 \cdot 78 = 2652$
    * $P_3 = (x_1 + x_0) \cdot (y_1 + y_0) = (12 + 34) \cdot (56 + 78) = 46 \cdot 134 = 6164$


3. **Compute middle term:**
    * $P_3 - P_1 - P_2 = 6164 - 672 - 2652 = 2840$


4. **Combine terms:**
    * $x \cdot y = P_1 \cdot 10^4 + (P_3 - P_1 - P_2) \cdot 10^2 + P_2$
    * $x \cdot y = 672 \cdot 10000 + 2840 \cdot 100 + 2652 = 6,720,000 + 284,000 + 2652 = 7,006,652$



---

# **3. Python Implementation**

```python
def karatsuba(x: int, y: int) -> int:
    # Base case for small numbers (fall back to standard multiplication)
    if x < 10 or y < 10:
        return x * y

    # Determine split size m
    n = max(len(str(x)), len(str(y)))
    m = n // 2

    # Split numbers into higher (x1, y1) and lower (x0, y0) parts
    high_x, low_x = divmod(x, 10**m)
    high_y, low_y = divmod(y, 10**m)

    # 3 Recursive calls
    p1 = karatsuba(high_x, high_y)
    p2 = karatsuba(low_x, low_y)
    p3 = karatsuba(high_x + low_x, high_y + low_y)

    # Recombine using Karatsuba identity
    return (p1 * 10 ** (2 * m)) + ((p3 - p1 - p2) * 10**m) + p2


# Sanity check
print(karatsuba(1234, 5678))  # Output: 7006652

```

# **4. Practical Thresholds & Applications**

While Karatsuba is asymptotically faster than grade-school multiplication, it carries overhead due to recursion and addition/subtraction steps.

* **Threshold crossover:** For small numbers (typically under 32–64 digits or 1024 bits), traditional grade-school multiplication is faster. Production BigInteger libraries (such as Python's CPython, Java's `BigInteger`, and GNU MP) use hybrid strategies:
    * $n < \text{Threshold}$: Grade-school multiplication ($\mathcal{O}(n^2)$)
    * $\text{Threshold} \le n \le \text{Upper Threshold}$: Karatsuba multiplication ($\mathcal{O}(n^{1.585})$)
    * $n > \text{Upper Threshold}$: Toom-Cook or Fast Fourier Transform (FFT / Schönhage–Strassen) ($\mathcal{O}(n \log n \log \log n)$)

**Time complexity** measures how the runtime of an algorithm grows as the size of the input ($n$) increases. Rather than measuring time in seconds (which varies based on hardware), time complexity counts the **number of elementary operations** executed by the algorithm.

---

# **1. Asymptotic Notations**

- We express time complexity using mathematical notations to describe upper, lower, and tight bounds:

```
    Runtime
       ▲
       │     /  O(f(n))  [Upper Bound - Worst Case]
       │    /
       │   /    Θ(f(n))  [Tight Bound - Average Case]
       │  /
       │ /      Ω(f(n))  [Lower Bound - Best Case]
       └──────────────────────────► Input Size (n)

```

1. **Big-O ($\mathcal{O}$): Upper Bound**
Represents the **worst-case scenario** (maximum operations). This is the standard notation used in computer science and software interviews because it provides a guaranteed performance upper limit.
2. **Big-Omega ($\Omega$): Lower Bound**
Represents the **best-case scenario** (minimum operations).
3. **Big-Theta ($\Theta$): Tight Bound**
Describes the exact growth rate when the best-case and worst-case bounds match.

# **2. Common Time Complexities Hierarchy**

- Ordered from fastest to slowest:

> $$\mathcal{O}(1) < \mathcal{O}(\log n) < \mathcal{O}(n) < \mathcal{O}(n \log n) < \mathcal{O}(n^2) < \mathcal{O}(2^n) < \mathcal{O}(n!)$$

| Complexity | Name | Example Algorithm / Operation | Description |
| --- | --- | --- | --- |
| **$\mathcal{O}(1)$** | Constant | Hash table lookup, Array index access | Runtime is independent of input size. |
| **$\mathcal{O}(\log n)$** | Logarithmic | Binary Search, Balanced Search Trees | Halves the search space at each step. |
| **$\mathcal{O}(n)$** | Linear | Linear Search, Array traversal | Runtime grows proportionally to input size. |
| **$\mathcal{O}(n \log n)$** | Linearithmic | Merge Sort, QuickSort (average case) | Standard bound for comparison-based sorting. |
| **$\mathcal{O}(n^2)$** | Quadratic | Bubble Sort, Nested loops over $n$ | Pairwise processing or nested comparisons. |
| **$\mathcal{O}(2^n)$** | Exponential | Recursive Fibonacci, Generating all subsets | Doubles with each additional element. |
| **$\mathcal{O}(n!)$** | Factorial | Traveling Salesperson (brute force), Permutations | Tries every possible ordering/combination. |

# **3. How to Calculate Time Complexity (Code Analysis)**

### **Rule 1: Drop Constants & Non-Dominant Terms**

* Big-O focuses on the growth rate as $n \to \infty$.
    * $5n^2 + 100n + 5000 \implies \mathcal{O}(n^2)$
    * $2n + 5 \implies \mathcal{O}(n)$

### **Rule 2: Sequential Steps Add Up**

```python
# Step 1: O(n)
for i in range(n):
    print(i)

# Step 2: O(n^2)
for i in range(n):
    for j in range(n):
        print(i, j)

# Total: O(n + n^2) => O(n^2)

```

### **Rule 3: Nested Loops Multiply**

```python
# Outer loop runs n times
for i in range(n):
    # Inner loop runs log(n) times (doubling)
    j = 1
    while j < n:
        j *= 2

# Total Time Complexity: O(n * log n) = O(n log n)

```

**Big O notation** ($\mathcal{O}$) is a mathematical framework used in computer science to describe the **upper bound** (worst-case scenario) of an algorithm's execution time or memory footprint relative to the input size $n$.

It focuses purely on **growth rates** as $n \to \infty$, ignoring hardware differences, CPU cycles, and low-level execution constants.

# **1. Formal Mathematical Definition**

- A function $f(n)$ is said to be $\mathcal{O}(g(n))$ if there exist positive constants $c$ and $n_0$ such that:
    - $$f(n) \le c \cdot g(n) \quad \text{for all } n \ge n_0$$

> This means that beyond a threshold input size $n_0$, the total operation count $f(n)$ will never exceed $g(n)$ scaled by a factor $c$.

```
     Operations
        ▲
        │           / c * g(n)  (Upper Bound)
        │          /
        │    _---_---- f(n)     (Actual Algorithm Execution)
        │  _/   /
        │ /    /
        ├─────/───────────────────► Input Size (n)
        0    n_0
```

# **3. Rules for Simplifying Big O Expressions**
- When deriving the Big O complexity of a code block or function:

### **1. Drop Non-Dominant Terms**

- Focus on the term that dominates as $n$ approaches infinity.
    * $f(n) = 3n^3 + 500n^2 + 10000 \implies \mathcal{O}(n^3)$
    * $f(n) = n + \log n \implies \mathcal{O}(n)$

### **2. Drop Constants**

- Scale factors do not change asymptotic growth families.
    * $f(n) = 500 \cdot n \implies \mathcal{O}(n)$
    * $f(n) = \frac{1}{2}(n^2 + n) \implies \mathcal{O}(n^2)$

### **3. Additive vs. Multiplicative Rules**

* **Sequential operations add:**

$$\mathcal{O}(A + B)$$


* **Nested operations multiply:**

$$\mathcal{O}(A \cdot B)$$



```python
# Additive: O(A + B)
for x in list_A:  # O(A)
    pass
for y in list_B:  # O(B)
    pass

# Multiplicative: O(A * B)
for x in list_A:  # O(A)
    for y in list_B:  # O(B)
        pass
```

## **4. Big O vs. Other Asymptotic Notations**

* **Big O ($\mathcal{O}$):** Upper bound ($f(n) \le c \cdot g(n)$) $\to$ **Worst Case**
* **Big Omega ($\Omega$):** Lower bound ($f(n) \ge c \cdot g(n)$) $\to$ **Best Case**
* **Big Theta ($\Theta$):** Tight bound ($c_1 \cdot g(n) \le f(n) \le c_2 \cdot g(n)$) $\to$ **Exact Growth Rate**

> **Example (QuickSort):**
> * Worst Case: $\mathcal{O}(n^2)$ (sorted input with bad pivot choice)
> * Best Case: $\Omega(n \log n)$
> * Average Case: $\Theta(n \log n)$

While **Big O ($\mathcal{O}$)** provides an upper bound on an algorithm's growth rate, **Big Omega ($\Omega$)** and **Big Theta ($\Theta$)** round out asymptotic analysis by defining the lower and tight bounds.

---

# **1. Summary Comparison**

```
                  Runtime
                     ▲
                     │            / c_2 * g(n)  [Big O: Upper Bound]
                     │           /
                     │     _---_---- f(n)       [Big Theta: Tight Bound]
                     │    /     /
                     │   /     / c_1 * g(n)     [Big Omega: Lower Bound]
                     └──/─────/─────────────────► Input Size (n)
                       0     n_0
```

* **Big Omega ($\Omega$):** "The algorithm takes **at least** this much time." ($\ge$)
* **Big Theta ($\Theta$):** "The algorithm takes **exactly** this order of magnitude time." ($=$)
* **Big O ($\mathcal{O}$):** "The algorithm takes **at most** this much time." ($\le$)

---

# **2. Big Omega Notation ($\Omega$)**

### **Definition**
- Big Omega defines the **formal lower bound** of an algorithm's growth rate. A function $f(n) = \Omega(g(n))$ if there exist positive constants $c$ and $n_0$ such that:
    - $$f(n) \ge c \cdot g(n) \quad \text{for all } n \ge n_0$$

### **Purpose**
- It guarantees that an algorithm will **never run faster** than a certain threshold for sufficiently large inputs.

### **Key Examples**

* **Comparison-Based Sorting:** Any comparison-based sorting algorithm (Merge Sort, QuickSort, Heap Sort) requires $\Omega(n \log n)$ comparisons in the worst case. You cannot break this theoretical lower bound without non-comparison techniques (like Counting Sort or Radix Sort).
* **Array Search:** Finding an element in an unsorted array of size $n$ takes $\Omega(1)$ in the best case (element is at index 0).

---

# **3. Big Theta Notation ($\Theta$)**

### **Definition**
- Big Theta defines a **tight bound** (both upper and lower). A function $f(n) = \Theta(g(n))$ if there exist positive constants $c_1$, $c_2$, and $n_0$ such that:
    - $$c_1 \cdot g(n) \le f(n) \le c_2 \cdot g(n) \quad \text{for all } n \ge n_0$$

### **Mathematical Equivalent**
- An algorithm is $\Theta(g(n))$ if and only if it is **both** $\mathcal{O}(g(n))$ and $\Omega(g(n))$ for the same $g(n)$:
    - $$f(n) = \Theta(g(n)) \iff f(n) = \mathcal{O}(g(n)) \quad \text{AND} \quad f(n) = \Omega(g(n))$$

### **Purpose**
- It provides the precise order of growth when the best-case and worst-case bounds match asymptotically.

### **Key Examples**
* **Merge Sort:** Always divides arrays and merges them regardless of initial order.
* Best case: $\Omega(n \log n)$
* Worst case: $\mathcal{O}(n \log n)$
* **Tight bound:** $\Theta(n \log n)$
* **Array Traversal:** Iterating through a $1D$ array of size $n$ always takes linear time: $\Theta(n)$.

| **$\mathcal{O}(g(n))$** | Big O | Upper Bound | $f(n) \le c \cdot g(n)$ | Worst-case scenario |
| **$\Omega(g(n))$** | Big Omega | Lower Bound | $f(n) \ge c \cdot g(n)$ | Best-case scenario / Problem lower bound |
| **$\Theta(g(n))$** | Big Theta | Tight Bound | $c_1 \cdot g(n) \le f(n) \le c_2 \cdot g(n)$ | Exact growth rate |

| **Merge Sort** | $\Omega(n \log n)$ | $\mathcal{O}(n \log n)$ | **Yes:** $\Theta(n \log n)$ |

**Amortized analysis** is a technique used in computer science to calculate the **average time per operation over a sequence of operations**, guaranteeing a tight bound even when individual operations occasionally incur high costs.

Unlike standard worst-case analysis (which assumes *every* step takes the maximum possible time) or average-case analysis (which relies on probability distributions), amortized analysis guarantees the average performance in the **worst-case sequence**.

---

# **1. Why Amortized Analysis Matters**

- Consider appending an element to a dynamic array (e.g., Python `list` or C++ `std::vector`):
    * **Normal insertion:** $1$ operation ($\mathcal{O}(1)$).
    * **Resizing step:** When full, the array doubles its capacity. It allocates a new block of memory and copies all $k$ existing elements ($\mathcal{O}(k)$ time).

If we only evaluated the worst single operation, we would claim appending takes $\mathcal{O}(n)$. However, array doubling happens infrequently ($1, 2, 4, 8, 16, \dots$). Amortized analysis proves that over $n$ appends, the total cost is $\mathcal{O}(n)$, making the **amortized cost per append $\mathcal{O}(1)$**.

# **2. The Three Methods of Amortized Analysis**

### **Method 1: Aggregate Method**

- Calculate the total cost $T(n)$ for a sequence of $n$ operations and divide by $n$:
    - $$\text{Amortized Cost} = \frac{T(n)}{n}$$

* **Dynamic Array Example ($n$ insertions starting from capacity 1):**
    * Insertion cost: $n$ operations ($1$ per element).
    * Copying cost during resizes at sizes $1, 2, 4, 8, \dots, 2^k$:
        - $$1 + 2 + 4 + 8 + \dots + 2^k < 2n$$
    * Total cost $T(n) < n + 2n = 3n \implies \mathcal{O}(n)$.
    * Amortized cost per operation: $\frac{3n}{n} = \mathbf{\mathcal{O}(1)}$.

---

### **Method 2: Accounting (Banker's) Method**

- Assign artificial "credits" or "charges" ($\hat{c}_i$) to each operation:
    * If $\hat{c}_i > c_i$ (actual cost), store the excess as **credit**.
    * If $\hat{c}_i < c_i$, use accumulated credit to pay for the expensive operation.
        - $$\sum_{i=1}^{n} \hat{c}_i \ge \sum_{i=1}^{n} c_i \quad \text{must hold for all } n$$
    * **Dynamic Array Example:**
        * Charge **$3** per append operation:
            * **$1** pays for inserting the current element.
            * **$1** saved as credit for copying this element during the next resize.
            * **$1** saved as credit to pay for copying an older element that has already moved.
        * When a resize occurs, every element has sufficient accumulated credit to pay for its copy operation.
        * Amortized cost per append: **$\mathcal{O}(1)$**.

---

### **Method 3: Potential Method (Physicist's Method)**

Define a **potential function** $\Phi(S)$ mapping the data structure state $S$ to a real number ($\Phi(S) \ge \Phi(S_0)$ for all states).

- The amortized cost $\hat{c}_i$ of the $i$-th operation is:
    - $$\hat{c}_i = c_i + \Phi(S_i) - \Phi(S_{i-1})$$

- Where $c_i$ is actual cost and $\Delta \Phi_i = \Phi(S_i) - \Phi(S_{i-1})$ is the change in potential energy:
    * If potential increases ($\Delta \Phi > 0$), work is saved for later.
    * If potential decreases ($\Delta \Phi < 0$), saved work pays for an expensive operation.

# **3. Classic Examples in Data Structures**

| Data Structure | Operation | Worst-Case Cost | Amortized Cost | Mechanism |
| --- | --- | --- | --- | --- |
| **Dynamic Array** | `append()` | $\mathcal{O}(n)$ | $\mathcal{O}(1)$ | Array doubling capacity |
| **Splay Tree** | `search()` / `insert()` | $\mathcal{O}(n)$ | $\mathcal{O}(\log n)$ | Splaying moves accessed node to root |
| **Fibonacci Heap** | `decrease_key()` | $\mathcal{O}(n)$ | $\mathcal{O}(1)$ | Lazy structural updates |
| **Disjoint Set (Union-Find)** | `find()` | $\mathcal{O}(n)$ | $\mathcal{O}(\alpha(n))$ | Path compression + rank/size heuristic |
