In Python, every line of code is either an **expression** or a **statement**. Understanding the difference between them is crucial for mastering syntax, functional patterns, writing clean code, and understanding how Python executes your program.

---

# **1. Expressions vs. Statements**

### **What is an Expression?**
- An **expression** is a combination of values, variables, operators, and function calls that **evaluates to a single value**.

- Because an expression produces a value, it can be assigned to a variable, passed as an argument to a function, or combined into larger expressions.

```python
# Examples of Expressions
42                         # Evaluates to integer: 42
3.14 + 2.0                 # Evaluates to float: 5.14
a > 10                     # Evaluates to boolean: True or False
len("Data Science")        # Evaluates to integer: 12
[x**2 for x in range(3)]   # Evaluates to list: [0, 1, 4]
```

### **What is a Statement?**

- A **statement** is an instruction or command that Python executes to **perform an action**. Statements do not produce or evaluate to a value on their own. Instead, they control execution flow, create variables, define functions, or manage program structure.

```python
# Examples of Statements
x = 10                     # Assignment statement (creates binding)
if x > 5:                  # Conditional control flow statement
    print("Greater")
for i in range(3):         # Loop statement
    pass
def greet():               # Function definition statement
    return "Hello"
```

# **2. Comparison Summary**

| Feature | Expression | Statement |
| --- | --- | --- |
| **Primary Goal** | Computes and **evaluates to a value**. | Executes an **action or command**. |
| **Returns a Value?** | Yes (always returns a result). | No (does not evaluate to a value). |
| **Can be on the right side of `=`?** | Yes (e.g., `y = x + 5`). | No (e.g., `y = (if x > 5: ...)` is invalid). |
| **Can be passed to functions?** | Yes (e.g., `print(a + b)`). | No (e.g., `print(import math)` is invalid). |
| **Examples** | `a + b`, `x == 2`, `func()`, `x if c else y` | `x = 5`, `if`, `for`, `def`, `class`, `import` |

# **3. Important Nuances & Edge Cases**

### **Expression Statements**

- In Python, an expression can stand alone on its own line. When it does, Python evaluates it and discards the result. This is called an **expression statement**.

```python
# Function call evaluates to "Data Science", but standing alone it acts as an expression statement:
len("Data Science") 

# In an interactive terminal (REPL or Jupyter), standalone expressions automatically print their result.
```

### **Statement vs. Expression Variants**

- Python often provides an *expression equivalent* for certain *statements*:

1. **`if/else` Statement vs. Ternary Expression:**
```python
# Statement (controls flow, no direct returned value)
if age >= 18:
    status = "Adult"
else:
    status = "Minor"

# Expression (evaluates directly to a value)
status = "Adult" if age >= 18 else "Minor"
```


2. **Function Definition Statement vs. Lambda Expression:**
```python
# Statement (defines a function object in namespace)
def square(x):
    return x ** 2

# Expression (evaluates to an anonymous function object inline)
square = lambda x: x ** 2
```


3. **Assignment Statement vs. The Walrus Operator (`:=`):**
- Standard variable assignment (`x = 5`) is a statement and cannot be evaluated inline. Python 3.8+ introduced the **Assignment Expression** (Walrus Operator `:=`), which assigns a value *and* returns that value inline.
```python
# Statement: Invalid syntax inside an 'if' condition
# if (n = len(data)) > 0:  # SyntaxError!

# Expression: Valid inline assignment
data = [1, 2, 3, 4]
if (n := len(data)) > 0:
    print(f"List contains {n} items.")
```

# **4. Why This Distinction Matters in Data Structures & Algorithms**

1. **List / Dict / Set Comprehensions:** Comprehensions require **expressions** inside their syntax, not statements.
```python
# Valid: Expression inside comprehension
squares = [x**2 for x in range(10) if x % 2 == 0]

# Invalid: You cannot put statements (like print or assignments) inside a comprehension
# items = [x = x + 1 for x in range(5)]  # SyntaxError!
```


2. **Recursion & Functional Programming:** Functions like `map()`, `filter()`, and key extractors in `sorted()` or `heapq` require callable expressions (like lambdas or function references) to process data efficiently.
3. **Time & Space Impact:** While both execute code, expressions evaluated inside loops (e.g., repeatedly calling `len(lst)` vs. assigning `n = len(lst)` once via statement) can impact time complexity if an expression is unintentionally expensive ($O(n)$ instead of $O(1)$).

# **4. Essential Extras & Algorithmic Nuances**

### **1. The Concatenation Trap ($O(n^2)$ vs $O(n)$)**

- Appending strings inside a loop using `+=` repeatedly creates and destroys intermediate string objects in memory, leading to an accidental $O(n^2)$ time complexity.

```python
# SLOW: O(n^2) time due to reallocating memory on each iteration
result = ""
for char in ["a", "b", "c", "d"]:
    result += char

# FAST: O(n) time using list collect + join
chars = ["a", "b", "c", "d"]
result = "".join(chars)  # Allocates memory once at the end

```

### **2. String Encoding (ASCII vs. UTF-8)**

- In Python 3, all `str` objects are sequences of Unicode code points.
    * To convert a string to raw bytes for network transmission or file storage, use `.encode('utf-8')` ($\rightarrow$ `bytes`).
    * To convert bytes back to a string, use `.decode('utf-8')`.

- To write efficient algorithms, it is crucial to understand how Python lists are built:

# **5. Essential Extras & Algorithmic Nuances**

### **1. The Queue Anti-Pattern (`lst.pop(0)`)**

Using a Python list as a Queue (First-In, First-Out) by calling `lst.pop(0)` leads to an **$O(n^2)$ total time runtime** when processing $n$ elements due to repeated pointer shifting.

```python
# SLOW: O(n) per pop(0) -> O(n^2) overall
q = [1, 2, 3, 4, 5]
while q:
    item = q.pop(0)  # Shifts all elements left!

# FAST: Use collections.deque (Doubly Linked List / Blocked Ring Buffer)
from collections import deque

q = deque([1, 2, 3, 4, 5])
while q:
    item = q.popleft()  # O(1) time operation!
```

### **2. Deep Copy vs. Shallow Copy**

Because lists store references, copying a list using `new_lst = lst` only copies the reference, not the data.

```python
import copy

a = [[1, 2], [3, 4]]

shallow_copy = a.copy()   # or a[:]
deep_copy = copy.deepcopy(a)

a[0][0] = 999
print(shallow_copy[0][0]) # 999 (Nested list reference was shared!)
print(deep_copy[0][0])    # 1 (Completely independent object created)
```

### **3. List Comprehensions vs Loops**

List comprehensions are not only more concise but also **execute faster** in Python than explicit `for` loops with `.append()`, because the bytecode loop is optimized in C inside the CPython interpreter.

```python
# Slower
squares = []
for x in range(1000):
    squares.append(x ** 2)

# Faster & More Pythonic
squares = [x ** 2 for x in range(1000)]
```

### **Feature 1: Assigning Functions to Variables**

- Since functions are objects, you can assign them to variables without invoking them (omit the `()` parentheses).

```python
def greet(name):
    return f"Hello, {name}!"

# Assign function object to a new variable name
say_hello = greet  

# Call the function via the new reference
print(say_hello("Alice"))  # Output: Hello, Alice!

```

### **Feature 2: Passing Functions as Arguments (Higher-Order Functions)**

- A function that accepts another function as an argument is called a **Higher-Order Function**.

```python
def apply_operation(func, numbers):
    """Applies a given function to each element in a list."""
    return [func(x) for x in numbers]

def square(n):
    return n ** 2

nums = [1, 2, 3, 4]
# Pass 'square' as a first-class object argument
results = apply_operation(square, nums)  # Output: [1, 4, 9, 16]

```

### **Feature 3: Returning Functions from Functions & Closures**

- Functions can construct and return inner function objects dynamically.

```python
def make_multiplier(factor):
    """Outer function returns an inner function."""
    def multiply(number):
        return number * factor  # Captures 'factor' from outer scope (Closure)
    
    return multiply  # Return function object reference

double = make_multiplier(2)  # double is now a function object
triple = make_multiplier(3)  # triple is another function object

print(double(5))  # Output: 10
print(triple(5))  # Output: 15
```

### **Feature 4: Storing Functions in Data Structures**

- Functions can be stored in lists, tuples, or dictionaries, enabling dynamic dispatch tables.

```python
def add(a, b): return a + b
def subtract(a, b): return a - b

# Dictionary mapping operations to function objects
operations = {
    "+": add,
    "-": subtract,
    "*": lambda a, b: a * b  # Lambda expression evaluates to a function object
}

# Dynamic function lookup and execution
result = operations["+"](10, 5)  # Executes add(10, 5) -> 15
```

### **2. Higher-Order Functions in Real-World Code: Decorators**

- Decorators in Python are higher-order functions that wrap another function to extend its behavior dynamically:

```python
import time

def timer_decorator(func):
    """Higher-order decorator measuring execution time."""
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        end = time.perf_counter()
        print(f"{func.__name__} took {end - start:.6f} seconds.")
        return result
    return wrapper

@timer_decorator
def compute_sum():
    return sum(range(1_000_000))

compute_sum()  # Automatically logs execution time!
```

A **Recursive Function** is a function that **calls itself** in order to solve a problem by breaking it down into smaller, self-similar subproblems.

Recursion is closely tied to mathematical induction and is the core technique behind advanced algorithmic paradigms like **Divide and Conquer** (e.g., Merge Sort, Quick Sort), **Dynamic Programming**, and **Tree/Graph Traversals**.

---

# **1. Code Breakdown & Structure**

- Every valid recursive function **must** contain two essential components:
    1. **Base Case(s):** The stopping condition that returns a direct value without making further recursive calls (prevents infinite loops).
    2. **Recursive Step:** The logic that reduces the problem size and calls the function again towards the base case.

### **Example: Calculating Factorial ($n!$)**

```python
def factorial(n):
    # 1. Base Case: 0! and 1! are explicitly defined as 1
    if n <= 1:
        return 1
    
    # 2. Recursive Step: n! = n * (n - 1)!
    return n * factorial(n - 1)

print(factorial(5))  # Output: 120
```

### **Trace of Execution (Call Stack)**

When calling `factorial(3)`, Python pushes each call onto the **Call Stack** until it hits the base case, then pops them off as values return:

```text
CALL STACK (Push Phase)            RETURN VALUE (Pop Phase)
-----------------------            ------------------------
factorial(3)                       Returns 3 * 2 = 6
  └─> 3 * factorial(2)             Returns 2 * 1 = 2
        └─> 2 * factorial(1)       Returns 1
              └─> Hits Base Case (n = 1)

```

# **2. Core Concepts: Memory & The Call Stack**

- Unlike iterative loops (`for`/`while`) which execute in a single stack frame, **each recursive call creates a new stack frame** in system memory to store:
    * Local variables.
    * Current parameter values.
    * The execution return address.

### **The Stack Overflow Problem (RecursionError)**

# **3. Time & Space Complexity Analysis**

- Analysing recursive algorithms requires evaluating both the **tree depth** and the **work done per recursive call**.

### **Comparison: Naive Recursion vs. Memoized Recursion (Fibonacci)**

#### **Naive Fibonacci ($O(2^n)$ Time)**

```python
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)  # Generates a binary call tree
```

* **Time Complexity:** $O(2^n)$ — Exponential explosion of redundant subproblem computations (e.g., `fib(3)` is recalculated repeatedly).
* **Space Complexity:** $O(n)$ — Memory used by the call stack at maximum depth $n$.

#### **Memoized Fibonacci with `functools.lru_cache` ($O(n)$ Time)**

- By storing the results of subproblems using Python's higher-order decorator `lru_cache`, we eliminate redundant recursive branches:

```python
from functools import lru_cache

@lru_cache(maxsize=None)
def fib_memoized(n):
    if n <= 1:
        return n
    return fib_memoized(n - 1) + fib_memoized(n - 2)

print(fib_memoized(50))  # Runs instantly!
```

* **Time Complexity:** $O(n)$ — Each value from $0$ to $n$ is calculated exactly once.
* **Space Complexity:** $O(n)$ — Stack depth plus $O(n)$ cache storage.

# **1. Generators (Data Producers)**

- A **generator** is a function that produces a sequence of values lazily (one at a time, on demand) rather than computing them all at once and storing them in memory.

### **Code Breakdown: Generator Function vs. List**

```python
def count_up_to(max_val):
    """Generator function that produces numbers up to max_val."""
    count = 1
    while count <= max_val:
        yield count  # Pauses execution and yields 'count' to caller
        count += 1   # Resumes here when next() is called

# Instantiating the generator object (does NOT execute code yet)
counter = count_up_to(3)

# Pulling values lazily
print(next(counter))  # Output: 1 (pauses at yield 1)
print(next(counter))  # Output: 2 (pauses at yield 2)
print(next(counter))  # Output: 3 (pauses at yield 3)

# Calling next() again raises StopIteration exception
# print(next(counter))  # StopIteration error
```

### **Generator Expressions**

- Just as list comprehensions create lists, **generator expressions** create generator objects using parentheses `()` instead of brackets `[]`:

```python
# List comprehension: Computes ALL 1,000,000 squares in RAM immediately
list_squares = [x**2 for x in range(1_000_000)]  # ~8 MB memory

# Generator expression: Computes squares ON DEMAND, 1 at a time
gen_squares = (x**2 for x in range(1_000_000))   # ~100 bytes memory
```

# **2. Coroutines (Data Consumers & Async Tasks)**

- While generators primary **push data out** (via `yield`), **coroutines** can also **receive data in** (via `value = yield`) or cooperate in asynchronous event loops.

### **Classic Generator-Based Coroutines (`.send()`)**

- By putting `yield` on the right side of an assignment (`value = yield`), a function becomes a consumer that receives values passed into it via the `.send()` method:

```python
def running_average():
    """Coroutine that calculates a running average of incoming numbers."""
    total = 0.0
    count = 0
    average = None
    
    while True:
        # Pauses and receives data sent via .send()
        term = yield average  
        if term is None:
            break
        total += term
        count += 1
        average = total / count

# 1. Instantiate coroutine
avg = running_average()

# 2. "Prime" the coroutine to advance execution to the first yield statement
next(avg)  # Returns initial 'average' (None)

# 3. Send data into the coroutine
print(avg.send(10))  # Output: 10.0
print(avg.send(20))  # Output: 15.0
print(avg.send(30))  # Output: 20.0
```

### **Modern Native Coroutines (`async` / `await`)**

- In modern Python (Python 3.5+), coroutines are primarily implemented using `async def` and `await` to run asynchronous code non-blockingly inside an event loop (e.g., `asyncio`):

```python
import asyncio

async def fetch_data(data_id):
    """Native coroutine function."""
    print(f"Fetching item {data_id}...")
    await asyncio.sleep(1)  # Yields control back to event loop without blocking thread
    return f"Data {data_id}"

async def main():
    # Run multiple coroutines concurrently
    results = await asyncio.gather(fetch_data(1), fetch_data(2))
    print(results)

# Run the event loop
# asyncio.run(main())
```

### **1. Delegating Generators (`yield from`)**

`yield from` allows a generator to delegate part of its operations to another sub-generator or iterable cleanly:

```python
def flatten(nested_list):
    """Flattens a 2D list using yield from."""
    for sublist in nested_list:
        yield from sublist  # Equivalent to: for item in sublist: yield item

matrix = [[1, 2], [3, 4], [5, 6]]
print(list(flatten(matrix)))  # Output: [1, 2, 3, 4, 5, 6]
```

A **representation error** occurs primarily when dealing with floating-point numbers (`float`). It happens because computer hardware represents numbers in binary (base-2), whereas humans represent numbers in decimal (base-10). Certain base-10 fractions **cannot be represented exactly** as finite binary fractions.

# **1. Code Breakdown & Demonstration**

### **The Infamous `0.1 + 0.2 != 0.3` Error**

```python
# Expectation: 0.3
# Reality: 0.30000000000000004
result = 0.1 + 0.2
print(result)

# Evaluates to False due to representation error!
print(0.1 + 0.2 == 0.3)  # Output: False
```

- You can inspect the exact stored value using `.hex()` or `decimal.Decimal`:

```python
from decimal import Decimal

# Shows the actual 64-bit IEEE 754 float representation in base-10
print(Decimal(0.1))
# Output: 0.1000000000000000055511151231257827021181583404541015625

print(Decimal(0.2))
# Output: 0.200000000000000011102230246251565404236316680908203125
```

---

# **2. Core Concept: Why Does Representation Error Happen?**

### **Base-10 vs. Base-2 Fractions**

- In decimal (base-10), a fraction can be written as a finite decimal if its denominator's prime factors are only $2$ and $5$ (the prime factors of $10$).
    * $\frac{1}{2} = 0.5$ (Finite)
    * $\frac{1}{5} = 0.2$ (Finite)
    * $\frac{1}{3} = 0.333333...$ (Infinite repeating fraction)

- In binary (base-2), a fraction can only be written as a finite binary string if its denominator's prime factors are **only $2$**.
    * $\frac{1}{2} = 0.1_2$ (Exact)
    * $\frac{1}{4} = 0.01_2$ (Exact)
    * $\frac{1}{10} = \frac{1}{2 \times 5} = 0.00011001100110011..._2$ (Infinite repeating sequence!)

### **IEEE 754 Floating-Point Standard**

- Python uses standard C-level 64-bit double-precision floating-point numbers:
    * **1 bit** for sign
    * **11 bits** for exponent
    * **52 bits** for mantissa (precision)

- Because precision is capped at 52 bits, Python must **round** the infinitely repeating binary fraction of $0.1$ to fit into 64 bits of memory, introducing a tiny approximation error (around $10^{-17}$).

# **3. How to Handle Representation Errors in Python**

- Depending on your algorithmic use case, Python provides three core techniques to handle or eliminate floating-point error:

### **Solution 1: Use `math.isclose()` for Equality Checks**

- Never compare float values using `==`. Instead, check if two numbers are "close enough" within a small tolerance (epsilon).

```python
import math

a = 0.1 + 0.2
b = 0.3

# Correct float comparison
print(math.isclose(a, b))  # Output: True
```

### **Solution 2: Use `decimal.Decimal` for Exact Base-10 Calculations**

The `decimal` module implements fixed and floating-point arithmetic using base-10 rather than base-2. It is essential for **financial calculations** or accounting.

```python
from decimal import Decimal

# Pass numbers as STRINGS to prevent binary conversion before Decimal creation
d1 = Decimal('0.1')
d2 = Decimal('0.2')

exact_sum = d1 + d2
print(exact_sum)         # Output: 0.3
print(exact_sum == Decimal('0.3'))  # Output: True
```

### **Solution 3: Use `fractions.Fraction` for Exact Rational Numbers**

- The `fractions` module stores numbers as explicit pairs of integers (numerator and denominator).

```python
from fractions import Fraction

f1 = Fraction(1, 10)  # 1/10
f2 = Fraction(2, 10)  # 2/10

print(f1 + f2)  # Output: 3/10
print(f1 + f2 == Fraction(3, 10))  # Output: True
```

# **3. How Dictionaries Work Under the Hood (Hash Tables)**

- Understanding Python's dictionary internals explains its performance characteristics:
    - 1. **The Hash Function (`hash()`):** When you execute `d[key] = val`, Python calls `hash(key)` to generate an integer hash value.
    - 2. **Index Computation:** Python maps this integer hash to an index slot in an internal lookup array.
    - 3. **Collision Handling:** If two keys hash to the same slot (a *hash collision*), Python resolves it using **Open Addressing with Quadratic Probing/Perturbation**.
    - 4. **Order Preservation (Python 3.7+):** Python dictionaries preserve **insertion order** while maintaining $O(1)$ lookup performance. This is achieved using a dual-array design: a dense array storing entries in insertion order, and a sparse hash array storing index offsets.

An **Abstract Data Type (ADT)** is a theoretical model that defines a data structure solely by **what operations it performs** and **what behavior it exhibits**, completely hiding **how those operations are implemented** in code or memory.

- In simple terms:
    * **ADT (The Interface / Contract):** Describes *what* can be done (e.g., "I need a container where I can add items and remove the most recently added item").
    * **Data Structure (The Concrete Implementation):** Describes *how* it is built in code using underlying memory structures (e.g., implementing that behavior using a contiguous array or a doubly linked list).

    * **The Concrete Data Structure:** The actual mechanical engine, transmission, and wiring under the hood that executes the acceleration.

# **3. Implementing an ADT in Python: The Stack ADT Example**

In Python, ADTs can be formally defined using **Abstract Base Classes (ABCs)** from the `abc` module. This enforces a strict contract that any concrete data structure must satisfy.

### **Defining the Stack ADT Interface**

```python
from abc import ABC, abstractmethod

class StackADT(ABC):
    """Abstract Data Type specifying the Stack contract."""
    
    @abstractmethod
    def push(self, item):
        """Add an item to the top of the stack."""
        pass

    @abstractmethod
    def pop(self):
        """Remove and return the top item from the stack."""
        pass

    @abstractmethod
    def peek(self):
        """Return the top item without removing it."""
        pass

    @abstractmethod
    def is_empty(self) -> bool:
        """Check if the stack is empty."""
        pass

```

### **Concrete Implementation 1: Dynamic Array Stack**

```python
class ArrayStack(StackADT):
    """Concrete implementation of Stack ADT using a Python list (Dynamic Array)."""
    
    def __init__(self):
        self._data = []

    def push(self, item):
        self._data.append(item)  # Amortized O(1)

    def pop(self):
        if self.is_empty():
            raise IndexError("pop from empty stack")
        return self._data.pop()  # O(1)

    def peek(self):
        if self.is_empty():
            raise IndexError("peek from empty stack")
        return self._data[-1]    # O(1)

    def is_empty(self) -> bool:
        return len(self._data) == 0

```

### **Concrete Implementation 2: Linked List Stack**

```python
class Node:
    def __init__(self, val, next=None):
        self.val = val
        self.next = next

class LinkedListStack(StackADT):
    """Concrete implementation of Stack ADT using a Singly Linked List."""
    
    def __init__(self):
        self._head = None
        self._size = 0

    def push(self, item):
        self._head = Node(item, self._head)  # O(1)
        self._size += 1

    def pop(self):
        if self.is_empty():
            raise IndexError("pop from empty stack")
        val = self._head.val
        self._head = self._head.next  # O(1)
        self._size -= 1
        return val

    def peek(self):
        if self.is_empty():
            raise IndexError("peek from empty stack")
        return self._head.val  # O(1)

    def is_empty(self) -> bool:
        return self._head is None

```

# **4. Why ADTs Matter in Software Engineering & Algorithms**

1. **Encapsulation and Decoupling:** Client code using an ADT doesn't care how the data structure is built under the hood. You can replace an `ArrayStack` with a `LinkedListStack` without breaking any external code calling `.push()` or `.pop()`.
2. **Algorithm Design:** Algorithms are typically designed around ADTs rather than specific memory implementations (e.g., "Dijkstra's Shortest Path algorithm requires a Priority Queue ADT").
3. **Flexibility to Optimize:** If performance requirements change (e.g., needing $O(1)$ guaranteed time instead of amortized $O(1)$), you swap the underlying data structure while keeping the same interface intact.

In Python, a **`deque`** (short for **Double-Ended Queue**, pronounced "deck") is a specialized data structure provided by the standard `collections` module.

While Python lists are optimized for random access by index, `deque` is specifically designed for **$O(1)$ fast appends and pops from both ends**.

---

# **1. Code Breakdown & Usage**

A `deque` can be used as a **Queue (FIFO)**, a **Stack (LIFO)**, or a **Bounded Buffer**.

```python
from collections import deque

# 1. Initialize a deque (optionally populated from an iterable)
dq = deque([10, 20, 30])

# 2. Adding Elements (Both Ends)
dq.append(40)        # Right side: deque([10, 20, 30, 40])
dq.appendleft(5)     # Left side:  deque([5, 10, 20, 30, 40])

# 3. Removing Elements (Both Ends)
right_val = dq.pop()      # Removes & returns 40 -> deque([5, 10, 20, 30])
left_val  = dq.popleft()  # Removes & returns 5  -> deque([10, 20, 30])

# 4. Peeking at Ends
first_item = dq[0]   # 10
last_item  = dq[-1]  # 30

# 5. Rotating Elements
# Positive n shifts right (clockwise); Negative n shifts left
dq.rotate(1)   # deque([30, 10, 20])
dq.rotate(-1)  # deque([10, 20, 30])
```

---

# **2. Core Concepts: Why Use `deque` Over `list`?**

### **The Python List Limitation ($O(n)$ Shifts)**

- Python lists are **dynamic contiguous arrays**.
    * Appending/popping from the right end (`lst.append()`, `lst.pop()`) takes $O(1)$ amortized time.
    * Inserting/popping from the left end (`lst.insert(0, val)`, `lst.pop(0)`) requires shifting **every single remaining element in memory to the left or right**, causing an **$O(n)$ time complexity penalty**.

### **How `deque` Works Under the Hood**

- In CPython, `deque` is implemented as a **doubly linked list of fixed-size block arrays** (typically 62 elements per block):
    * Pointers maintain references to the head and tail blocks.
    * Appending or popping from either end operates in **strict $O(1)$ time** without needing to reallocate or shift the entire collection.

---

# **3. Comparison: `list` vs. `deque`**

| Feature | `list` (Dynamic Array) | `collections.deque` (Doubly Linked Blocks) |
| --- | --- | --- |
| **Append / Pop (Right)** | Amortized $O(1)$ | Strict $O(1)$ |
| **Append / Pop (Left)** | **$O(n)$** *(Slow: Element Shifting)* | **Strict $O(1)$** *(Ultra Fast)* |
| **Random Access by Index (`d[i]`)** | **$O(1)$** *(Direct Pointer Math)* | **$O(n)$** *(Must traverse internal block list)* |
| **Insert / Delete at Middle** | $O(n)$ | $O(n)$ |
| **Memory Layout** | Single contiguous memory block | Memory allocated in linked blocks |

---

# **4. Time & Space Complexity Analysis**

| Operation | Time Complexity | Space Complexity | Reason / Explanation |
| --- | --- | --- | --- |
| `append(x)` / `appendleft(x)` | $O(1)$ | $O(1)$ | Inserts at head/tail block buffer. |
| `pop()` / `popleft()` | $O(1)$ | $O(1)$ | Removes from head/tail block buffer. |
| `rotate(k)` | $O(k)$ | $O(1)$ | Moves $k$ pointers internally across blocks. |
| Index Access (`dq[i]`) | $O(n)$ | $O(1)$ | Traverses block chain to index $i$ (optimized to search from nearest end). |
| Full Iteration (`for x in dq`) | $O(n)$ | $O(n)$ | Sequential traversal across linked blocks. |

In Python, a **`ChainMap`** is a specialized container provided by the standard `collections` module. It allows you to **group multiple dictionaries (or mappings) into a single, updateable view** without copying their data or physically merging them into a new dictionary.

When you look up a key in a `ChainMap`, it searches through the underlying dictionaries sequentially, returning the value from the **first mapping that contains the key**.

---

# **1. Code Breakdown & Demonstration**

```python
from collections import ChainMap

# Define three separate dictionaries (e.g., configuration layers)
defaults = {"theme": "light", "show_sidebar": True, "font_size": 12}
user_settings = {"theme": "dark", "font_size": 14}
cmd_args = {"show_sidebar": False}

# Create a ChainMap search chain (order defines precedence: Left-to-Right)
config = ChainMap(cmd_args, user_settings, defaults)

# 1. Searching/Lookup (Checks cmd_args -> user_settings -> defaults)
print(config["theme"])         # 'dark' (Found in user_settings)
print(config["show_sidebar"])  # False  (Found in cmd_args)
print(config["font_size"])     # 14     (Found in user_settings)

# 2. Key Mutability & Operations
# ALL writes, updates, and deletes affect ONLY the FIRST dictionary in the chain!
config["font_size"] = 16  # Updates cmd_args!

print(cmd_args)      # {'show_sidebar': False, 'font_size': 16}
print(user_settings) # {'theme': 'dark', 'font_size': 14} (Unchanged!)
```

### **Context Management Methods (`new_child` and `parents`)**

`ChainMap` provides utilities to push or pop scoping layers dynamically—ideal for programming language scopes or context stacks:

```python
# Create a new child scope (prepends a new empty dict at the front)
nested_config = config.new_child({"theme": "high-contrast"})

print(nested_config["theme"])  # 'high-contrast' (From new scope)

# Access parent scopes (skips the first dictionary)
parent_config = nested_config.parents
print(parent_config["theme"])  # 'dark'
```

---

# **2. Core Concepts: Why Use `ChainMap` instead of `dict.update()`?**

### **Comparison: `dict.update()` vs. `ChainMap`**

```python
# APPROACH A: dict.update() or {**d1, **d2}
# - Creates a BRAND NEW merged dictionary in memory (O(N) copy cost).
# - Loses original dictionary independence (changes to 'defaults' won't reflect).
merged = defaults.copy()
merged.update(user_settings)
merged.update(cmd_args)

# APPROACH B: ChainMap
# - Zero data copying (O(1) creation time).
# - Dynamic linkage: Changes in underlying dicts dynamically update the ChainMap!
chain = ChainMap(cmd_args, user_settings, defaults)
defaults["font_size"] = 18  # Instantly reflected in 'chain' if not overridden!
```

---

# **3. How `ChainMap` Works Under the Hood**

- Under the hood in Python:
    * `ChainMap` stores the mappings in an internal list attribute called `.maps`.
    * `config.maps` gives you direct access to the list of dictionaries: `[cmd_args, user_settings, defaults]`.
    * Key lookups iterate over `self.maps` and execute standard dictionary key checks ($O(1)$ per map) until a match is found.

---

# **4. Time & Space Complexity Analysis**

| Operation | Time Complexity | Space Complexity | Explanation |
| --- | --- | --- | --- |
| **Creation** `ChainMap(*maps)` | **$O(k)$** | **$O(k)$** | $k$ is the number of mapped dictionaries. Only stores references to dict objects (no item copying). |
| **Lookup** `cm[key]` | **$O(k)$** worst case | $O(1)$ | Checks each of the $k$ maps sequentially until found ($O(1)$ per map). |
| **Insertion / Update** `cm[key] = val` | $O(1)$ | $O(1)$ | Directly writes to `cm.maps[0]` in $O(1)$ time. |
| **Deletion** `del cm[key]` | $O(1)$ | $O(1)$ | Deletes key from `cm.maps[0]`. Raises `KeyError` if key isn't in the first map. |
| **Iteration** `for k in cm:` | $O(N)$ | $O(N)$ | Iterates over all $N$ unique keys across all $k$ maps. |

In Python, **`Counter`** is a specialized dictionary subclass provided by the standard `collections` module. It is designed specifically for **counting hashable objects**, tracking frequency distributions, and finding top elements without manual loop boilerplate.

---

# **1. Code Breakdown & Usage**

`Counter` transforms any iterable (strings, lists, tuples) or mapping into a frequency dictionary where items are stored as dictionary **keys** and their counts are stored as dictionary **values**.

```python
from collections import Counter

# 1. Initialization
c_str = Counter("abracadabra")
c_list = Counter(["apple", "banana", "apple", "cherry", "banana", "apple"])

print(c_list)
# Output: Counter({'apple': 3, 'banana': 2, 'cherry': 1})

# 2. Accessing Missing Keys (Returns 0 instead of raising KeyError!)
print(c_list["mango"])  # Output: 0

# 3. Finding Top Elements (Most Common)
# Returns a list of (element, count) tuples sorted by highest frequency
print(c_list.most_common(2))
# Output: [('apple', 3), ('banana', 2)]

# 4. Unpacking Elements Back to an Iterable
print(list(c_list.elements()))
# Output: ['apple', 'apple', 'apple', 'banana', 'banana', 'cherry']
```

---

# **2. Mathematical Set-like & Arithmetic Operations**

`Counter` objects support arithmetic and set operations for combining or comparing frequencies:

```python
c1 = Counter(a=3, b=1, c=0)
c2 = Counter(a=1, b=2, d=4)

# Addition: Sum counts from both counters
print(c1 + c2)  # Counter({'a': 4, 'b': 3, 'd': 4})

# Subtraction: Subtract counts (keeps ONLY positive counts!)
print(c1 - c2)  # Counter({'a': 2})

# Intersection (Min): Keeps minimum count of overlapping elements
print(c1 & c2)  # Counter({'a': 1, 'b': 1})

# Union (Max): Keeps maximum count for each element
print(c1 | c2)  # Counter({'a': 3, 'b': 2, 'd': 4})
```

### **Updating Counts Directly**

- Use `.update()` to increment counts or `.subtract()` to decrement them in-place:

```python
counts = Counter(["a", "b"])
counts.update(["a", "c"])    # Counter({'a': 2, 'b': 1, 'c': 1})
counts.subtract(["a", "a"])  # Counter({'a': 0, 'b': 1, 'c': 1})
```

---

# **3. How `Counter` Works Under the Hood**

- Under the hood in CPython:
    * `Counter` inherits directly from Python’s built-in `dict`.
    * Key lookups use standard **Hash Table lookup** ($O(1)$ average time).
    * Overridden `__getitem__` returns `0` for missing keys instead of raising a `KeyError`.
    * `.most_common(k)` uses **`heapq.nlargest`** when $k$ is small, taking $O(n \log k)$ time, or sorts the whole dictionary ($O(n \log n)$) when $k$ is large or unspecified.

# **1. Important Context: Python 3.7+ vs. `OrderedDict`**

- Since Python 3.7, built-in `dict` instances preserve insertion order as an official language specification. However, **`OrderedDict` is not obsolete**—it offers critical structural and algorithmic features that standard `dict` lacks:
    - 1. **Equality testing is order-sensitive:** Two `OrderedDict` instances with the same key-value pairs compare as **equal only if their keys appear in the exact same order**.
    - 2. **Reordering operations:** Includes built-in methods like `.move_to_end()` to reorder keys efficiently in $O(1)$ time.
    - 3. **Pop options:** `.popitem(last=True)` allows popping from either the end (LIFO) or the start (FIFO) in $O(1)$ time.

---

# **2. Code Breakdown & Operations**

```python
from collections import OrderedDict

# 1. Creation and Initialization
od = OrderedDict([('a', 1), ('b', 2), ('c', 3)])

# 2. Moving Elements to Ends in O(1)
od.move_to_end('a')               # Move 'a' to the far right (end)
# Result: OrderedDict([('b', 2), ('c', 3), ('a', 1)])

od.move_to_end('c', last=False)   # Move 'c' to the far left (beginning)
# Result: OrderedDict([('c', 3), ('b', 2), ('a', 1)])

# 3. Popping Elements (LIFO vs. FIFO)
last_item = od.popitem(last=True)    # Removes & returns ('a', 1) [Right side / LIFO]
first_item = od.popitem(last=False)  # Removes & returns ('c', 3) [Left side / FIFO]

# 4. Order-Sensitive Equality
od1 = OrderedDict([('a', 1), ('b', 2)])
od2 = OrderedDict([('b', 2), ('a', 1)])
print(od1 == od2)  # False! (Order differs)

d1 = {'a': 1, 'b': 2}
d2 = {'b': 2, 'a': 1}
print(d1 == d2)    # True! (Standard dict ignores order during equality)
```

---

# **3. How `OrderedDict` Works Under the Hood**

- Under the hood in CPython:
    * Standard `dict` uses a dense key-value array paired with an index table to preserve insertion order.
    * `OrderedDict` pairs a **hash table** with a **doubly linked list of keys**.
    * Each entry holds explicit pointers (`prev` and `next`) to adjacent keys. This is what enables **$O(1)$ reordering (`move_to_end`) and $O(1)$ popping from both ends (`popitem(last=False)`)**.
    * Due to these double pointers, `OrderedDict` consumes roughly **2x to 2.5x more memory** than a standard `dict`.

---

# **4. Time & Space Complexity Analysis**

| Operation | Time Complexity | Space Complexity | Explanation / Mechanism |
| --- | --- | --- | --- |
| **Lookup** `od[key]` | **$O(1)$** | $O(1)$ | Hash table lookup. |
| **Insertion** `od[key] = val` | **$O(1)$** | $O(1)$ | Inserts key into hash table and appends node to doubly linked list. |
| **Delete** `del od[key]` | **$O(1)$** | $O(1)$ | Unlinks node from doubly linked list in $O(1)$ time. |
| **`move_to_end(key)`** | **$O(1)$** | $O(1)$ | Adjusts doubly linked list pointers around the target node. |
| **`popitem(last=Bool)`** | **$O(1)$** | $O(1)$ | Removes head or tail node from doubly linked list. |
| **Memory Overhead** | — | **$O(n)$** | Requires space for double pointers (`prev`/`next`) for every key. |

# **5. Algorithmic Applications & Design Patterns**

### **1. LRU Cache (Least Recently Used Cache) Implementation**

- `OrderedDict` is the classic native Python structure used to build an **LRU Cache** in clean, minimal code:

```python
class LRUCache:
    def __init__(self, capacity: int):
        self.capacity = capacity
        self.cache = OrderedDict()

    def get(self, key: int) -> int:
        if key not in self.cache:
            return -1
        # Mark as recently used by moving it to the end
        self.cache.move_to_end(key)
        return self.cache[key]

    def put(self, key: int, value: int) -> None:
        if key in self.cache:
            self.cache.move_to_end(key)
        self.cache[key] = value
        
        # Evict least recently used item (first item) if full
        if len(self.cache) > self.capacity:
            self.cache.popitem(last=False)

```

In Python, **`defaultdict`** is a dictionary subclass provided by the standard `collections` module. It inherits all behavior from standard dictionaries, but automatically provides a **default value for a non-existent key** when accessed, eliminating the need to check for key existence or handle `KeyError` exceptions.

---

# **1. Code Breakdown & Usage**

`defaultdict` takes a **callable factory function** (such as `list`, `int`, `set`, or a `lambda`) as its first argument. When a missing key is accessed, `defaultdict` calls this factory without arguments, assigns the returned value to the key, and returns it.

```python
from collections import defaultdict

# 1. Grouping items into lists (factory: list -> default: [])
grouped_data = defaultdict(list)
pairs = [("fruit", "apple"), ("vegetable", "carrot"), ("fruit", "banana")]

for category, item in pairs:
    grouped_data[category].append(item)  # No KeyError if key doesn't exist!

print(dict(grouped_data))
# Output: {'fruit': ['apple', 'banana'], 'vegetable': ['carrot']}

# 2. Counting frequencies (factory: int -> default: 0)
counts = defaultdict(int)
for char in "abracadabra":
    counts[char] += 1  # 0 + 1 on first occurrence

print(dict(counts))
# Output: {'a': 5, 'b': 2, 'r': 2, 'c': 1, 'd': 1}

# 3. Custom default values via lambda
custom_dict = defaultdict(lambda: "N/A")
custom_dict["name"] = "Alice"

print(custom_dict["name"])    # Output: 'Alice'
print(custom_dict["status"])  # Output: 'N/A' (Key 'status' is created and set to 'N/A')
```

---

# **2. Key Behavior Nuance: Access Mutates the Dictionary**

A critical detail to remember: **merely reading or accessing a missing key in a `defaultdict` inserts that key into the dictionary with its default value**.

```python
d = defaultdict(list)

# Accessing a missing key triggers creation
print(d["missing_key"])  # Output: []

# 'missing_key' now exists in the dictionary!
print("missing_key" in d)  # Output: True
print(len(d))              # Output: 1

```

> **Tip:** If you only want to inspect a key without mutating or inserting defaults, use standard `.get()`:
> ```python
> value = d.get("another_missing", None)  # Does NOT insert key into d

# **4. How `defaultdict` Works Under the Hood**

- Under the hood in CPython:
    * `defaultdict` overrides the special `__missing__(key)` method defined in Python dictionary objects.
    * When `d[key]` fails to locate `key` in the hash table, Python automatically triggers `d.__missing__(key)`.
    * `__missing__(key)` checks `self.default_factory`. If it is `None`, it raises a standard `KeyError`. Otherwise, it calls `self.default_factory()`, stores the result in `self[key]`, and returns it.

---

# **5. Time & Space Complexity Analysis**

| Operation | Time Complexity | Space Complexity | Explanation |
| --- | --- | --- | --- |
| **Creation** `defaultdict(factory)` | **$O(1)$** | $O(1)$ | Assigns factory pointer in $O(1)$ time. |
| **Existing Key Lookup** `d[key]` | **$O(1)$** | $O(1)$ | Direct hash table lookup. |
| **Missing Key Access** `d[missing]` | **$O(1)$** | **$O(1)$** | Calls factory and inserts key-value into hash table. |
| **Iteration** `for k, v in d.items()` | **$O(n)$** | $O(n)$ | Standard hash table traversal over $n$ elements. |

# **6. Algorithmic Patterns & Real-World Use Cases**

### **1. Building Adjacency Lists for Graph Algorithms**

`defaultdict(list)` or `defaultdict(set)` is the standard Python pattern for graph representations:

```python
edges = [("A", "B"), ("A", "C"), ("B", "C"), ("C", "D")]
graph = defaultdict(list)

for u, v in edges:
    graph[u].append(v)
    graph[v].append(u)  # Undirected graph

# graph['A'] -> ['B', 'C']
```

### **2. Grouping Anagrams ($O(N \cdot K \log K)$ Time)**

```python
def group_anagrams(words: list[str]) -> list[list[str]]:
    groups = defaultdict(list)
    for word in words:
        sorted_key = "".join(sorted(word))
        groups[sorted_key].append(word)
    return list(groups.values())
```

In Python, **`namedtuple`** is a factory function provided by the standard `collections` module. It creates lightweight, immutable, tuple-like objects whose fields can be accessed via **named attribute lookups** (e.g., `point.x`) as well as standard integer indexing (e.g., `point[0]`).

---

# **1. Code Breakdown & Usage**

`namedtuple` returns a custom subclass of Python's built-in `tuple`. It gives you the low memory footprint of a tuple with the clarity and readabilty of an object or class.

```python
from collections import namedtuple

# 1. Define a namedtuple class (typename, field_names)
Point = namedtuple("Point", ["x", "y", "z"])

# 2. Instantiate objects
p1 = Point(x=10, y=20, z=30)
p2 = Point(1, 2, 3)  # Positional arguments work too!

# 3. Access fields
print(p1.x, p1.y)  # 10 20 (Attribute access)
print(p1[0], p1[1])  # 10 20 (Index access)

# 4. Tuple Unpacking & Iteration
x_val, y_val, z_val = p1
print(x_val, y_val, z_val)  # 10 20 30

# 5. Immutability
# p1.x = 100  # Raises AttributeError: can't set attribute
```

### **Useful Built-in Helper Methods**

- All `namedtuple` instances include special methods prefixed with a leading underscore to prevent collisions with field names:

```python
# Convert to dictionary
print(p1._asdict())  # {'x': 10, 'y': 20, 'z': 30}

# Create a new instance from a dict or iterable
data = {"x": 5, "y": 15, "z": 25}
p3 = Point._make([5, 15, 25])  # From iterable
p4 = Point(**data)  # Unpacking dict

# Replace fields (Returns a NEW instance, preserving immutability)
p5 = p1._replace(x=99)  # Point(x=99, y=20, z=30)

```

# **2. Type Hints Variant: `typing.NamedTuple`**

- For modern Python codebases using static type hinting, the class-based syntax provided by `typing.NamedTuple` is preferred:

```python
from typing import NamedTuple


class User(NamedTuple):
    id: int
    name: str
    email: str
    is_active: bool = True  # Supports default values!


user = User(id=1, name="Alice", email="alice@example.com")
print(user.name)  # 'Alice'
```

# **4. How `namedtuple` Works Under the Hood**

- Under the hood in Python:
    * `namedtuple()` dynamically generates Python class code as a string and evaluates it (`exec`) to create a new class inheriting from `tuple`.
    * Because it inherits directly from `tuple`, instances do **not** maintain a `__dict__` dictionary per object for storing attribute key-value pairs.
    * Attributes map directly to tuple offsets (indices) in C memory, giving it the exact same performance and memory characteristics as plain tuples.

In Python, the **`array`** module provides a compact, low-level data structure (the `array.array` type) for storing **homogenous, primitive C-style data types** (such as integers, floats, and characters).

While standard Python `list` objects store pointers to arbitrary Python objects scattered in memory, `array.array` stores pure binary data contiguously in memory—making it significantly more **memory-efficient** when dealing with millions of primitive values.

---

# **1. Code Breakdown & Usage**

When instantiating an `array.array`, you must pass a **typecode** as the first argument to define the underlying C data type.

```python
import array

# 1. Create an array of signed 32-bit integers (typecode 'i')
# Syntax: array.array(typecode, [initial_values])
numbers = array.array('i', [10, 20, 30, 40, 50])

# 2. Array operations (behaves like a mutable sequence/list)
numbers.append(60)          # Append item
numbers.extend([70, 80])    # Extend with iterable
numbers.insert(0, 5)        # Insert at index

print(numbers)              # Output: array('i', [5, 10, 20, 30, 40, 50, 60, 70, 80])

# 3. Direct Binary I/O (Fast File Operations)
# Write array contents directly to a binary file
with open("data.bin", "wb") as f:
    numbers.tofile(f)

# Read binary data back directly into an array
new_numbers = array.array('i')
with open("data.bin", "rb") as f:
    new_numbers.fromfile(f, 9)  # Read 9 integers

print(new_numbers)          # Output identical to original array
```

### **Type Constraints**

- Attempting to insert a value of a different data type causes a `TypeError`:

```python
ints = array.array('i', [1, 2, 3])
# ints.append(3.14)  # Raises TypeError: integer argument expected, got float
```

# **3. Core Concepts: Memory Layout (`list` vs. `array` vs. `numpy`)**

- Understanding why `array` exists requires looking at Python's C-level memory structures:

```text
STANDARD PYTHON LIST [10, 20, 30] (Array of Pointers)
+-------+-------+-------+
| PTR 1 | PTR 2 | PTR 3 |  --> Pointing to PyObject headers scattered in heap
+-------+-------+-------+
    |       |       |
    v       v       v
 [PyInt] [PyInt] [PyInt]  (Each PyObject wrapper takes ~28 bytes!)

ARRAY MODULE array('i', [10, 20, 30]) (Contiguous C Buffer)
+----+----+----+
| 10 | 20 | 30 |  --> Pure raw 4-byte integers stored back-to-back in RAM
+----+----+----+
```

### **Memory Benchmark Example**

- Storing 10 million 64-bit integers:
    * **Python `list`:** $\approx 80 \text{ MB}$ (Pointers array) $+ 280 \text{ MB}$ (Integer PyObjects) $\approx \mathbf{360 \text{ MB}}$
    * **Python `array('q')`:** $\approx \mathbf{80 \text{ MB}}$ total (Raw contiguous buffer)
